# THEORY QUESTIONS

QUESTION 1:
What is unsupervised learning in the context of machine learning?

ANSWER:
Unsupervised learning is a type of machine learning where the model learns patterns from unlabeled data without predefined output labels.


QUESTION 2:
How does K-Means clustering algorithm work?

ANSWER:
K-Means initializes k centroids, assigns each data point to the nearest centroid, updates centroids as the mean of assigned points, and repeats until convergence.


QUESTION 3:
Explain the concept of a dendrogram in hierarchical clustering.

ANSWER:
A dendrogram is a tree-like diagram that shows how clusters are merged or split at different distance levels in hierarchical clustering.


QUESTION 4:
What is the main difference between K-Means and Hierarchical Clustering?

ANSWER:
K-Means requires predefined k and uses centroid-based partitioning, while Hierarchical Clustering builds a tree structure without needing predefined cluster count.


QUESTION 5:
What are the advantages of DBSCAN over K-Means?

ANSWER:
DBSCAN can detect arbitrary shaped clusters and identify noise points, while K-Means assumes spherical clusters and cannot detect noise.


QUESTION 6:
When would you use Silhouette Score in clustering?

ANSWER:
Silhouette Score is used to evaluate how well data points fit within their clusters and to determine the optimal number of clusters.


QUESTION 7:
What are the limitations of Hierarchical Clustering?

ANSWER:
It is computationally expensive, not scalable to large datasets, and once clusters are merged or split, the process cannot be reversed.


QUESTION 8:
Why is feature scaling important in clustering algorithms like K-Means?

ANSWER:
K-Means relies on distance calculations, so unscaled features can dominate clustering results.


QUESTION 9:
How does DBSCAN identify noise points?

ANSWER:
DBSCAN labels points as noise if they do not have enough neighboring points within the specified radius.


QUESTION 10:
Define inertia in the context of K-Means.

ANSWER:
Inertia is the sum of squared distances between data points and their nearest cluster centroid.


QUESTION 11:
What is the elbow method in K-Means clustering?

ANSWER:
The elbow method plots inertia against number of clusters and identifies the point where the rate of decrease sharply changes.


QUESTION 12:
Describe the concept of density in DBSCAN.

ANSWER:
Density refers to the number of points within a specified radius. High density regions form clusters.


QUESTION 13:
Can hierarchical clustering be used on categorical data?

ANSWER:
Yes, if an appropriate distance metric such as Hamming distance is used.


QUESTION 14:
What does a negative Silhouette Score indicate?

ANSWER:
It indicates that data points may be assigned to the wrong cluster.


QUESTION 15:
Explain the term linkage criteria in hierarchical clustering.

ANSWER:
Linkage criteria define how distance between clusters is calculated such as single, complete, or average linkage.


QUESTION 16:
Why might K-Means perform poorly on varying cluster sizes or densities?

ANSWER:
Because it assumes equal-sized spherical clusters and minimizes variance.


QUESTION 17:
What are the core parameters in DBSCAN?

ANSWER:
eps defines neighborhood radius and min_samples defines minimum points to form a dense region.


QUESTION 18:
How does K-Means++ improve initialization?

ANSWER:
It selects initial centroids far apart from each other to improve convergence and clustering quality.


QUESTION 19:
What is agglomerative clustering?

ANSWER:
Agglomerative clustering is a bottom-up approach where each point starts as a separate cluster and merges iteratively.


QUESTION 20:
Why is Silhouette Score better than inertia?

ANSWER:
Silhouette Score measures both cohesion and separation, while inertia only measures compactness.

QUESTION:
Generate synthetic data with 4 centers using make_blobs and apply K-Means clustering. Visualize.

ANSWER:


```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=4, random_state=42)

kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], s=200)
plt.show()
```

QUESTION:
Evaluate silhouette scores for k=2 to 5 using make_blobs. Display bar chart.

ANSWER:


```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=4, random_state=42)

scores = []
ks = range(2,6)

for k in ks:
    labels = KMeans(n_clusters=k, random_state=42).fit_predict(X)
    scores.append(silhouette_score(X, labels))

plt.bar(ks, scores)
plt.show()
```

QUESTION:
Generate overlapping blobs, apply KMeans, visualize decision boundaries.

ANSWER:


```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=2.5, random_state=42)

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

h = .02
x_min, x_max = X[:,0].min()-1, X[:,0].max()+1
y_min, y_max = X[:,1].min()-1, X[:,1].max()+1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))

Z = kmeans.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:,0], X[:,1], c=kmeans.labels_)
plt.show()
```

QUESTION:
Load Breast Cancer dataset and compare inertia values for K = 2 to 6 using KMeans. Show line plot.

ANSWER:


```python
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X = load_breast_cancer().data

inertia = []
ks = range(2,7)

for k in ks:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X)
    inertia.append(model.inertia_)

plt.plot(ks, inertia, marker='o')
plt.xlabel("K")
plt.ylabel("Inertia")
plt.show()
```

QUESTION:
Load Wine dataset, apply KMeans, visualize cluster assignments using seaborn pairplot (first 4 features).

ANSWER:



```python
from sklearn.datasets import load_wine
from sklearn.cluster import KMeans
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = load_wine()
X = data.data[:, :4]

labels = KMeans(n_clusters=3, random_state=42).fit_predict(X)

df = pd.DataFrame(X, columns=data.feature_names[:4])
df["Cluster"] = labels

sns.pairplot(df, hue="Cluster")
plt.show()
```
